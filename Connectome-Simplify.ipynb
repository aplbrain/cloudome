```python
import networkx as nx
```

```python
RAW_DYNAMO_EXPORT = "./c40k/hannah-connectome-40k.csv"
```

```python
g = nx.MultiDiGraph()
with open(RAW_DYNAMO_EXPORT, "r") as f:
    # skip header
    for _ in range(1):
        next(f)
    for line in f:
        if line.startswith("#"):
            continue
        _, edge_raw = line.strip().split(",")
        # syn_x1000_y1068_z444_pre-1_post-1
        print(edge_raw)
        _, x, y, z, pre, post = edge_raw.split("_")
        x, y, z = int(x[1:]), int(y[1:]), int(z[1:])
        pre = pre[len("pre"):]
        post = post[len("post"):]
        g.add_edge(pre, post, pos=(x, y, z))
```

```python
# save the raw graph as a graphml
nx.write_edgelist(g, "./c40k/hannah-connectome-40k-raw.edgelist")
# remove -1 and 0 vertices
gg = g.copy()
gg.remove_node("-1")
gg.remove_node("0")
# save the graph as a graphml
nx.write_edgelist(gg, "./c40k/hannah-connectome-40k-processed.edgelist")
nx.write_edgelist(nx.DiGraph(gg), "./c40k/hannah-connectome-40k-simple.edgelist")
```

```python
print("Raw:", g.number_of_nodes(), g.number_of_edges())
print("No 0:", gg.number_of_nodes(), gg.number_of_edges())
simple = nx.DiGraph(gg)
print("Simple:", simple.number_of_nodes(), simple.number_of_edges())
print("% Reduction of edges (processed):", (g.number_of_edges() - gg.number_of_edges()) / g.number_of_edges())
print("% Reduction of edges (simple):", (g.number_of_edges() - simple.number_of_edges()) / g.number_of_edges())
```

```python
nx.density(g), nx.density(gg), nx.density(simple)
```
