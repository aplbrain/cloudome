```python
import networkx as nx
```

```python
contactome = nx.read_edgelist("./c40k/c40k_weighted.csv", delimiter=",", create_using=nx.Graph(), data=(("weight", float),))
```

```python
len(contactome.nodes()), len(contactome.edges())
```

```python
connectome = nx.read_edgelist("./c40k/hannah-connectome-40k-simple.edgelist", create_using=nx.Graph())
```

```python
len(connectome.nodes()), len(connectome.edges())
```

```python
list(contactome.nodes)[:10]
```

```python
list(connectome.nodes)[:10]
```

```python
# Size of set intersection of nodes
len(set(contactome.nodes()).intersection(connectome.nodes()))
```

```python
# How many (% and #) of contactome edges are in the connectome?
contactome_edges = set(contactome.edges())
connectome_edges = set(connectome.edges())

print(f"Contactome edges: {len(contactome_edges)}")
print(f"Connectome edges: {len(connectome_edges)}")
print(f"Contactome edges in connectome: {len(contactome_edges.intersection(connectome_edges))}")
print(f"Contactome edges in connectome (%): {len(contactome_edges.intersection(connectome_edges)) / len(contactome_edges) * 100:.2f}%")
print(f"Connectome edges in contactome: {len(connectome_edges.intersection(contactome_edges))}")
print(f"Connectome edges in contactome (%): {len(connectome_edges.intersection(contactome_edges)) / len(connectome_edges) * 100:.2f}%")
```

```python
subgraph_contactome = contactome.subgraph(connectome.nodes())
subgraph_connectome = connectome.subgraph(contactome.nodes())
```

```python
# How many (% and #) of contactome edges are in the connectome?
contactome_edges = set(subgraph_contactome.edges())
connectome_edges = set(subgraph_connectome.edges())

print(f"Subset Contactome edges: {len(contactome_edges)}")
print(f"Subset Connectome edges: {len(connectome_edges)}")
print(f"Subset Contactome edges in connectome: {len(contactome_edges.intersection(connectome_edges))}")
print(f"Subset Contactome edges in connectome (%): {len(contactome_edges.intersection(connectome_edges)) / len(contactome_edges) * 100:.2f}%")
print(f"Subset Connectome edges in contactome: {len(connectome_edges.intersection(contactome_edges))}")
print(f"Subset Connectome edges in contactome (%): {len(connectome_edges.intersection(contactome_edges)) / len(connectome_edges) * 100:.2f}%")
```
